```python
import matplotlib as plt
import numpy as np
import array 
```

```python
# Linear regression dataset: Years of Experience (X) vs. Salary in thousands (Y)

Years_of_Experience= np.array([1.1, 1.3, 1.5, 2.0, 2.2, 2.9, 3.0, 3.2, 3.2, 3.7, 3.9, 4.0, 4.0, 4.1, 4.5, 4.9, 5.1, 5.3, 5.9, 6.0])

Salary= np.array([39.3, 46.2, 37.7, 43.5, 39.8, 56.6, 60.1, 54.4, 64.4, 57.1, 63.2, 55.7, 56.9, 57.0, 61.1, 67.9, 66.0, 83.0, 81.3, 93.9])

#Initializing the parameters

Intercept = 0.0
slope = 0.0
hypothesis = 0.0


```

```python
# hypothesis = slope * years_of_experience + intercept
intercept = 30
slope = 1

#print(len(Years_of_Experience))    
#print(len(Salary)) 

result= np.array([])
for i in range(len(Salary)):
    hypothesis = slope*Years_of_Experience[i] + intercept
    print(hypothesis)
    result = np.append(result, hypothesis)


```

```python
import matplotlib.pyplot as plt

# 1. Plot the actual data points as a scatter plot
plt.scatter(Years_of_Experience, Salary, color='blue', label='Actual Data')

# 2. Plot your hypothesis line using the results array you just calculated
plt.plot(Years_of_Experience, result, color='red', linewidth=3, label='Hypothesis (slope=1, intercept=30)')

# 3. Add labels and a title
plt.xlabel('Years of Experience')
plt.ylabel('Salary (in thousands)')
plt.title('Linear Regression: Data vs Hypothesis Line')
plt.legend()
plt.grid(True)

# 4. Display the graph
plt.show()
```

```python
#Cost function

cost =0.0
m = len(Salary)
cost = sum((np.array(Salary) - np.array(result))**2) / 2*m

print("Cost of the hypothesis function: ", cost)
```

```python
#Gradient Descent
#temp intercept and slop initialization
learning_rate = 0.01
temp_intercept = intercept
temp_slope = slope

# 1. Calculate the gradients (partial derivatives)

temp_intercept = (1 / m) * sum(result - Salary)
temp_slope = (1 / m) * np.sum((result - Salary) * np.array(Years_of_Experience))

# 2. Update the model parameters simultaneously
intercept = intercept - (learning_rate * temp_intercept)
slope = slope - (learning_rate * temp_slope)
```

```python
# Convert your variables to numpy arrays once before starting
X_arr = np.array(Years_of_Experience)
Y_arr = np.array(Salary)
result = np.array(result)  # Using your pre-calculated old result array

learning_rate = 0.01
iterations = 1000

for i in range(iterations):
    
    temp_intercept = (1 / m) * np.sum(result - Y_arr)
    temp_slope = (1 / m) * np.sum((result - Y_arr) * X_arr)
    
    # 2. Calculate the step size changes (Delta values)
    step_intercept = learning_rate * temp_intercept
    step_slope = learning_rate * temp_slope
    
    # 3. Update parameters
    intercept = intercept - step_intercept
    slope = slope - step_slope
    
    # 4. Update your old 'result' array directly using the changes!
    # (Since result = slope*X + intercept, its change is exactly change_in_slope*X + change_in_intercept)
    result = result - (step_slope * X_arr + step_intercept)
```
